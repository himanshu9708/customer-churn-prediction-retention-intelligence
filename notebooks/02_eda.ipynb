# Phase 4 — Exploratory Data Analysis
EDA on the cleaned customer churn dataset. Findings are descriptive associations, not causal conclusions.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/customer_churn_cleaned.csv')
df.shape
```

```python
df.describe(include='all').T
```

```python
df['Churn'].value_counts(normalize=True).sort_index()
```

```python
numeric_features = ['Age','Tenure','Usage Frequency','Support Calls','Payment Delay','Total Spend','Last Interaction']
df[numeric_features + ['Churn']].corr(numeric_only=True)['Churn'].sort_values(ascending=False)
```

```python
for column in ['Gender','Subscription Type','Contract Length']:
    display(df.groupby(column)['Churn'].mean().sort_values(ascending=False))
```

## Key observations
- Churn is 56.71% in the cleaned dataset.
- Support Calls has the strongest positive numeric association with Churn.
- Total Spend has the strongest negative numeric association with Churn.
- Monthly contracts have 100% observed churn and require data-quality/business-rule investigation.
- These are associations, not causal effects.
