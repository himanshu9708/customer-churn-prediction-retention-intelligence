# Phase 7 — ML Model Development
Compare Logistic Regression, Random Forest, and XGBoost using a stratified holdout and 5-fold cross-validation.

```python
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold

df = pd.read_csv('../data/processed/customer_churn_features.csv')
X = df.drop(columns=['Churn'])
y = df['Churn'].astype(int)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, stratify=y, random_state=42)
X_train.shape, X_test.shape
```

## Models
The production training script in `src/models/train_models.py` builds preprocessing pipelines and trains Logistic Regression, Random Forest, and XGBoost.

```python
from sklearn.metrics import roc_auc_score, average_precision_score, recall_score, precision_score, f1_score

print('Train churn rate:', y_train.mean())
print('Test churn rate:', y_test.mean())
```

## Selection principle
Prefer PR-AUC and recall alongside ROC-AUC/F1 rather than selecting a model using accuracy alone. Final threshold tuning belongs after model comparison and should be based on business retention costs.
