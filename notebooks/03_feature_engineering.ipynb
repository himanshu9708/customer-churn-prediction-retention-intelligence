# Phase 6 — Feature Engineering
Create leakage-safe features. `CustomerID` is an identifier and `Churn` is the target.

```python
import pandas as pd

df = pd.read_csv('../data/processed/customer_churn_cleaned.csv')
df.shape
```

```python
df['support_calls_per_tenure'] = df['Support Calls'] / df['Tenure'].clip(lower=1)
df['usage_per_tenure'] = df['Usage Frequency'] / df['Tenure'].clip(lower=1)
df['spend_per_tenure'] = df['Total Spend'] / df['Tenure'].clip(lower=1)
df.head()
```

```python
df['tenure_band'] = pd.cut(df['Tenure'], [-float('inf'),6,12,24,48,float('inf')], labels=['0-6','7-12','13-24','25-48','49+'])
df['support_call_band'] = pd.cut(df['Support Calls'], [-float('inf'),2,5,8,float('inf')], labels=['0-2','3-5','6-8','9+'])
df['payment_delay_band'] = pd.cut(df['Payment Delay'], [-float('inf'),0,7,14,30,float('inf')], labels=['0','1-7','8-14','15-30','31+'])
df['usage_frequency_band'] = pd.cut(df['Usage Frequency'], [-float('inf'),5,10,20,30,float('inf')], labels=['0-5','6-10','11-20','21-30','31+'])
```

```python
X = df.drop(columns=['CustomerID','Churn'])
y = df['Churn']
X.shape, y.shape
```
